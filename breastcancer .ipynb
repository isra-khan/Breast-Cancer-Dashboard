```python
bins = [29, 40, 50, 60, 69]
labels = ["<=40", "41-50", "51-60", "61-69"]

df["age_group"] = pd.cut(df["age"], bins=bins, labels=labels, include_lowest=True)
df["age_group"].value_counts(dropna=False)
```

```python
df["tumor_size_category"] = pd.cut(
    df["tumor_size"],
    bins=[0, 20, 50, float("inf")],
    labels=["Small", "Medium", "Large"]
)
```

```python
df["positive_node_ratio"] = (
    df["reginol_node_positive"] /
    df["regional_node_examined"].replace(0, np.nan)
)
```

```python
df["hormone_receptor_status"] = (
    df["estrogen_status"] + "_" + df["progesterone_status"]
)
```

```python
df["lymph_node_involvement"] = np.where(
    df["reginol_node_positive"] > 0,
    "Yes",
    "No"
)
df = df.drop(columns=["lymph_node_involvement"])
```

```python
df["grade"] = df["grade"].astype(str).str.strip()
df["grade"] = df["grade"].replace({"anaplastic; Grade IV": "4"})
df["grade"].value_counts()
```

```python
# ==========================================
# PART 6 — SURVIVAL DURATION ANALYSIS
# ==========================================

# 1. Categorize Survival Months

df["survival_group"] = pd.cut(
    df["survival_months"],
    bins=[0, 12, 24, 60, 120, float("inf")],
    labels=[
        "0-12 Months",
        "13-24 Months",
        "25-60 Months",
        "61-120 Months",
        "120+ Months"
    ]
)

# 2. Number of Patients in Each Survival Group

print("Patients by Survival Duration")

print(df["survival_group"].value_counts().sort_index())

sns.countplot(
    data=df,
    x="survival_group",
    order=[
        "0-12 Months",
        "13-24 Months",
        "25-60 Months",
        "61-120 Months",
        "120+ Months"
    ]
)

plt.title("Patients by Survival Duration")
plt.xlabel("Survival Duration")
plt.ylabel("Number of Patients")
plt.xticks(rotation=30)
plt.show()


# 3. Percentage of Patients in Each Group

print("\nSurvival Duration Percentage")

print(
    df["survival_group"]
    .value_counts(normalize=True)
    .sort_index() * 100
)


# 4. Mean and Median Survival Months

print("\nSurvival Duration Statistics")

print("Mean:", df["survival_months"].mean())
print("Median:", df["survival_months"].median())
print("Minimum:", df["survival_months"].min())
print("Maximum:", df["survival_months"].max())


# 5. Survival Groups vs Patient Status

print("\nSurvival Duration vs Status")

print(
    pd.crosstab(
        df["survival_group"],
        df["status"]
    )
)

sns.countplot(
    data=df,
    x="survival_group",
    hue="status",
    order=[
        "0-12 Months",
        "13-24 Months",
        "25-60 Months",
        "61-120 Months",
        "120+ Months"
    ]
)

plt.title("Survival Duration Groups by Patient Status")
plt.xlabel("Survival Duration")
plt.ylabel("Number of Patients")
plt.xticks(rotation=30)
plt.show()
```

```python
# Step 8: Outlier Detection

# Important numerical columns
numerical_cols = [
    "age",
    "tumor_size",
    "survival_months"
]

# Boxplots for outlier visualization
for col in numerical_cols:
    plt.figure(figsize=(8, 3))
    sns.boxplot(x=df[col])
    plt.title(f"Outlier Detection - {col}")
    plt.xlabel(col)
    plt.show()


# IQR Method for identifying outliers
for col in numerical_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\nOutliers in {col}")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```
